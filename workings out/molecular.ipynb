```python
import pandas as pd
import re

train_path = "./X_train/molecular_train.csv"
test_path = "./X_test/molecular_test.csv"

def load_molecular(train_path, test_path):
    train = pd.read_csv(train_path)
    test = pd.read_csv(test_path)
    return train, test
```

```python
train, test = load_molecular(train_path, test_path)
train.head()
```
